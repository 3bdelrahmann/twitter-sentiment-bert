### CUSTOM DEFINED FUNCTIONS TO CLEAN THE TWEETS


```python
# Clean emojis from text
def remove_emojis(text):
    # Emoji pattern matching any emoji character
    emoji_pattern = re.compile("["
                               u"\U0001F600-\U0001F64F"  # emoticons
                               u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                               u"\U0001F680-\U0001F6FF"  # transport & map symbols
                               u"\U0001F700-\U0001F77F"  # alchemical symbols
                               u"\U0001F780-\U0001F7FF"  # Geometric Shapes Extended
                               u"\U0001F800-\U0001F8FF"  # Supplemental Arrows-C
                               u"\U0001F900-\U0001F9FF"  # Supplemental Symbols and Pictographs
                               u"\U0001FA00-\U0001FA6F"  # Chess Symbols, Symbols and Misc. Symbols
                               u"\U0001FA70-\U0001FAFF"  # Symbols and Pictographs Extended-A
                               u"\U00002600-\U000027BF"  # Miscellaneous Symbols
                               u"\U000024C2-\U0001F251" 
                               "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)

# Remove punctuations, links, mentions and \r\n new line characters
def clean_text(text):
    if text is None:
        return ''  # Return an empty string if input text is None

    # Convert to string and replace \r and \n with a space, then convert to lowercase
    text = str(text).replace('\r', ' ').replace('\n', ' ').lower()

    # Remove links (http/https) and mentions (@username)
    text = re.sub(r"(?:\@|https?\://)\S+", "", text)

    # Remove non-ASCII characters
    text = re.sub(r'[^\x00-\x7F]+', '', text)

    # Define a list of specific banned characters to remove
    banned_list = string.punctuation + 'Ã' + '±' + 'ã' + '¼' + 'â' + '»' + '§'

    # Create a translation table to remove banned characters
    table = str.maketrans('', '', banned_list)

    # Apply the translation table to clean the text
    text = text.translate(table)

    return text

# Clean hashtags at the end of the sentence, and keep those in the middle of the sentence by removing just the # symbol
def clean_hashtags(tweet):
    new_tweet = " ".join(word.strip() for word in re.split('#(?!(?:hashtag)\b)[\w-]+(?=(?:\s+#[\w-]+)*\s*$)', tweet)) #remove last hashtags
    new_tweet2 = " ".join(word.strip() for word in re.split('#|_', new_tweet)) #remove hashtags symbol from words in the middle of the sentence
    return new_tweet2

# Filter special characters such as & and $ present in some words
def filter_chars(a):
    sent = []
    for word in a.split(' '):
        if ('$' in word) | ('&' in word):
            sent.append('')
        else:
            sent.append(word)
    return ' '.join(sent)

def remove_mult_spaces(text): # remove multiple spaces
    return re.sub("\s\s+" , " ", text)
```

```python
# Define a function to preprocess a single tweet text
def preprocess_tweet(tweet_text):
    # Step 1: Remove emojis
    cleaned_text = remove_emojis(tweet_text)
    
    # Step 2: Remove punctuations, links, mentions, and newline characters
    cleaned_text = clean_text(cleaned_text)
    
    # Step 3: Clean hashtags
    cleaned_text = clean_hashtags(cleaned_text)
    
    # Step 4: Filter special characters such as '$' and '&'
    cleaned_text = filter_chars(cleaned_text)
    
    # Step 5: Remove multiple spaces
    cleaned_text = remove_mult_spaces(cleaned_text)
    
    return cleaned_text
```

```python
# Apply the preprocessing function to each tweet in the 'OriginalTweet' column
texts_new = [preprocess_tweet(t) for t in df['OriginalTweet']]

# Apply the preprocessing function to 'OriginalTweet' column in df_test
texts_new_test = [preprocess_tweet(t) for t in df_test['OriginalTweet']]
```

### Enhancing the Cleaning Process for the training Data

```python
# Initialize BERT tokenizer with 'bert-base-multilingual-cased' model
tokenizer = BertTokenizerFast.from_pretrained('bert-base-multilingual-cased')
```

```python
# Initialize an empty list to store token lengths
tokens_length = []

# Iterate over each text in the 'text_clean' column of the DataFrame 'df'
for txt in df['text_clean'].values:
    # Tokenize the text using the BERT tokenizer
    # - `tokenizer.encode`: Tokenizes the text and converts it into token IDs
    # - `max_length=512`: Specifies the maximum length of the tokenized sequence
    # - `truncation=True`: Truncates the sequence if it exceeds the maximum length
    tokens = tokenizer.encode(txt, max_length=512, truncation=True)
    
    # Append the length of the tokenized sequence to the list 'tokens_length'
    tokens_length.append(len(tokens))

# Calculate the maximum token length across all tokenized sequences
max_len = np.max(tokens_length)

# Print the maximum tokenized sentence length
print(f"Max Tokenized Sentence Length: {max_len}")
```

```python
# Initialize an empty list to store token lengths
tokens_length = []

# Iterate over each text in the 'text_clean' column of the DataFrame 'df' with its index
for i, txt in enumerate(df['text_clean'].values):
    # Tokenize the text using the BERT tokenizer with a maximum length of 512 tokens and truncation
    tokens = tokenizer.encode(txt, max_length=512, truncation=True)
    
    # Append the length of the tokenized sequence to the list 'tokens_length'
    tokens_length.append(len(tokens))
    
    # Check if the length of the tokenized sequence exceeds 80 tokens
    if len(tokens) > 80:
        # Print the index and text content for sequences longer than 80 tokens
        print(f"INDEX: {i}, TEXT: {txt}")

# Convert the list of token lengths to a NumPy array
tokens_length = np.array(tokens_length)
```

### Class Balancing by RandomOverSampler

```python
# Instantiate the RandomOverSampler
ros = RandomOverSampler()

# Reshape input features ('text_clean') and target ('Sentiment') for RandomOverSampler
X = np.array(df['text_clean']).reshape(-1, 1)  # Reshape text_clean as a 2D array
y = np.array(df['Sentiment'])                  # Use Sentiment directly as 1D array

# Apply RandomOverSampler to balance the classes
X_resampled, y_resampled = ros.fit_resample(X, y)

# Create a DataFrame from the resampled data
train_os = pd.DataFrame({
    'text_clean': [x[0] for x in X_resampled],  # Extract text_clean from the reshaped X_resampled
    'Sentiment': y_resampled                    # Use the resampled Sentiment (y_resampled)
})
```

```python
# Calculate and display the value counts of each sentiment label in the 'Sentiment' column
sentiment_counts = train_os['Sentiment'].value_counts()
print(sentiment_counts)
```

```python
# Extract input features (X) and target variable (y) from train_os DataFrame
X = train_os['text_clean'].values  # Extract 'text_clean' column as numpy array
y = train_os['Sentiment'].values   # Extract 'Sentiment' column as numpy array
```

```python
# Split data into training and validation sets
X_train, X_valid, y_train, y_valid = train_test_split(X, y, test_size=0.1, stratify=y, random_state=42)

# Display shapes of the resulting datasets 
print(f"Shape of X_train: {X_train.shape}")
print(f"Shape of X_valid: {X_valid.shape}")
print(f"Shape of y_train: {y_train.shape}")
print(f"Shape of y_valid: {y_valid.shape}")
```

```python
# Extract input features (X_test) and target variable (y_test) from df_test DataFrame
X_test = df_test['text_clean'].values  # Extract 'text_clean' column as numpy array
y_test = df_test['Sentiment'].values   # Extract 'Sentiment' column as numpy array
```

### Baseline model: Naive Bayes Classifier

```python
# Create CountVectorizer instance
clf = CountVectorizer()

# Fit and transform training data (X_train)
X_train_cv = clf.fit_transform(X_train)

# Transform test data (X_test) using the fitted CountVectorizer
X_test_cv = clf.transform(X_test)

# Display shapes of the transformed data
print(f"Shape of X_train_cv: {X_train_cv.shape}")
print(f"Shape of X_test_cv: {X_test_cv.shape}")
```

```python
# Create TfidfTransformer instance and fit on TF matrix (X_train_cv)
tf_transformer = TfidfTransformer(use_idf=True).fit(X_train_cv)

# Transform TF matrix of training data (X_train_cv) into TF-IDF matrix (X_train_tf)
X_train_tf = tf_transformer.transform(X_train_cv)

# Transform TF matrix of test data (X_test_cv) into TF-IDF matrix (X_test_tf)
X_test_tf = tf_transformer.transform(X_test_cv)

# Display shapes of the transformed data
print(f"Shape of X_train_tf: {X_train_tf.shape}")
print(f"Shape of X_test_tf: {X_test_tf.shape}")
```

```python
# Create Multinomial Naive Bayes classifier instance
nb_clf = MultinomialNB()

# Train the classifier using TF-IDF vectors (X_train_tf) and label-encoded target (y_train_le)
nb_clf.fit(X_train_tf, y_train_le)

# Use the trained classifier (nb_clf) to make predictions on the test data (X_test_tf)
nb_pred = nb_clf.predict(X_test_tf)
```

```python
from sklearn.metrics import classification_report

# Define target names corresponding to the class labels
target_names = ['Negative', 'Neutral', 'Positive']

# Generate and print the classification report for Naive Bayes classifier (nb_clf)
print('Classification Report for Naive Bayes:\n')
print(classification_report(y_test_le, nb_pred, target_names=target_names))
```

### BERT Sentiment Analysis

```python
def customTokenize(data, max_len=128):
    input_ids = []
    attention_masks = []
    for i in range(len(data)):
        encoded = tokenizer.encode_plus(
            data[i],
            add_special_tokens=True,
            max_length=max_len,  # Use the specified max_len
            padding='max_length',
            return_attention_mask=True
        )
        input_ids.append(encoded['input_ids'])
        attention_masks.append(encoded['attention_mask'])
    return np.array(input_ids), np.array(attention_masks)
```

```python
# Tokenize the training data
train_input_ids, train_attention_masks = customTokenize(X_train)

# Tokenize the validation data
val_input_ids, val_attention_masks = customTokenize(X_valid)

# Tokenize the test data
test_input_ids, test_attention_masks = customTokenize(X_test)
```

```python
# Initialize BERT model from 'bert-base-multilingual-cased'
bert_model = TFBertModel.from_pretrained('bert-base-multilingual-cased')
```

```python
def create_model(bert_model, max_len=128):
    # Define learning rate schedule parameters
    initial_learning_rate = 1e-5
    decay_steps = 1000
    decay_rate = 0.96

    # Define the learning rate schedule using ExponentialDecay
    lr_schedule = tf.keras.optimizers.schedules.ExponentialDecay(
        initial_learning_rate,
        decay_steps=decay_steps,
        decay_rate=decay_rate,
        staircase=True
    )

    # Define optimizer with the learning rate schedule
    opt = tf.keras.optimizers.Adam(learning_rate=lr_schedule)

    # Define loss and accuracy metrics
    loss = tf.keras.losses.CategoricalCrossentropy()
    accuracy = tf.keras.metrics.CategoricalAccuracy()

    # Define input layers for input_ids and attention_masks
    input_ids = tf.keras.Input(shape=(max_len,), dtype='int32')
    attention_masks = tf.keras.Input(shape=(max_len,), dtype='int32')

    # Obtain BERT embeddings using the BERT model
    embeddings = bert_model([input_ids, attention_masks])[1]  # Use pooled_output (index 1)

    # Add a Dense layer for classification (3 classes) with softmax activation
    output = tf.keras.layers.Dense(3, activation='softmax')(embeddings)

    # Define the model with inputs and output
    model = tf.keras.models.Model(inputs=[input_ids, attention_masks], outputs=output)

    # Compile the model with optimizer, loss, and metrics
    model.compile(optimizer=opt, loss=loss, metrics=[accuracy])

    return model
```

```python
# Create the text classification model using the create_model function
model = create_model(bert_model)

# Display the model summary
model.summary()
```

```python
# Train the model using model.fit() with training and validation data
history_bert = model.fit(
    [train_input_ids, train_attention_masks],  # Training input data (input_ids, attention_masks)
    y_train,  # Training labels
    validation_data=([val_input_ids, val_attention_masks], y_valid),  # Validation data
    epochs=4,  # Number of training epochs
    batch_size=32  # Batch size for training
)
```

```python
# Save the trained model to 'model/'
model.save('model/')
```

```python
# Extract training and validation loss from the history object
train_loss = history_bert.history['loss']
val_loss = history_bert.history['val_loss']

# Plot training and validation loss
plt.figure(figsize=(10, 6))
epochs = range(1, len(train_loss) + 1)
plt.plot(epochs, train_loss, 'b', label='Training Loss')
plt.plot(epochs, val_loss, 'r', label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)
plt.show()
```

```python
result_bert = model.predict([test_input_ids,test_attention_masks])
```

```python
y_pred_bert =  np.zeros_like(result_bert)
y_pred_bert[np.arange(len(y_pred_bert)), result_bert.argmax(1)] = 1
```
